## Introduction

Social networks are typically described as open platforms for global communication, but do users truly engage with others who have differing perspectives or are they confined within tightly connected circles of like-minded individuals?

In this project, I will analyze the Twitter network from the SNAP Ego-Twitter dataset, a large-scale graph capturing follower relationships between users. Unlike datasets that focus on content or engagement metrics, this network provides a structural view of how users are connected, which allows us to study the underlying topology of interactions at scale.

Our objective is to move beyond simple connectivity metrics to uncover the structural patterns that govern how users cluster and interact. Specifically, we investigate:

  - **The Existence of Echo Chambers:**

    Can unsupervised graph-based methods identiify tightly connected groups without prior labeling?

  - **Structural Isolation V.S Connectivity:**

    Do these communities exhibit dense internal connections with limited external links?

  - **The Role of Bridge Users**

    Are there users who act as connectors between otherwise disconnected communities?







## Research Strategy

For this notebook, I follow a research strategy that is tailored to solving the research question

### Phase 1 - Structural Context
- I characterize the graph globally, computing for attributes such as size, sparsity, reciprocity, components, and degree behavior.

### Phase 2 - Unsupervised Discovery of Communities
- I apply unsupervised community-detection methods and examine whether they recover meaningful user groups.

### Phase 3 - Testing Structural Isolation
- I evaluate whether the recovered communities are internally dense and externally sparse using multiple metrics: modularity, internal/external edge concentration, conductance, and internal density

### Phase 4 - Bridge User Analysis
- I identify users whose neighborhood spans multiple communities, and I interpret their role in enabling cross-community flow.

### Phase 5 - Robustness and Validation
- I compare the observed graph to a degree-preserving randomized baseline to test whether the discovered community structure reflects real organization rather than only degree sequence effects.

## Data Loading

The SNAP Ego-Twitter dataset is distributed as an edge list. I load it into a directed NetworkX graph, where an edge from `u` to `v` indicates a follower relationship. This is useful for context and descriptive analysis. As we progress further into the notebook, I will convert the directed graph into an undirected one to gather insight into group cohesion.

```python
# ------------------------------------------------------
# Download Directed Graph
# ------------------------------------------------------
# Source Page: https://snap.stanford.edu/data/ego-Twitter.html
# File name on SNAP: https://snap.stanford.edu/data/twitter_combined.txt.gz

# download dataset
DATA_URL = "https://snap.stanford.edu/data/twitter_combined.txt.gz"
LOCAL_PATH = "data/twitter_combined.txt.gz"

os.makedirs("data", exist_ok=True)

def download_dataset(url: str, to_path: str):
    import urllib.request
    print(f"Downloading from {url} ...")
    urllib.request.urlretrieve(url, to_path)
    size = os.path.getsize(to_path) / (1024*1024)
    print(f"Saved to {to_path} ({size:.2f} MB)")

download_dataset(DATA_URL, LOCAL_PATH)
```

```python
# ------------------------------------------------------
# Load Directed Graph
# ------------------------------------------------------
import gzip

def load_directed_graph(path, comment='#', sep=None) -> nx.DiGraph:
  '''
  Load directed edge list into a DiGraph
  Assumes there are comment lines
  Assumes lines are formatted like: u<sep>v
  '''
  graph = nx.DiGraph()

  # line counters
  line_count = 0
  skipped_lines_count = 0

  open_fn = gzip.open if path.endswith('.gz') else open

  with open_fn(path, 'rt') as f:
    for line in f:
      line_count += 1

      # ignore lines that start with # (comments)
      if line.startswith(comment):
        skipped_lines_count += 1
        continue

      components = line.strip().split(sep)

      # ignore lines that are malformed
      if len(components) != 2:
        skipped_lines_count += 1
        continue

      try:
        src, dst = map(int, components)
        graph.add_edge(src, dst)
      except ValueError:
        skipped_lines_count += 1
        continue

    return graph

print('Loading graph...')
graph = load_directed_graph(LOCAL_PATH)
print('Graph loaded.')
```

# Phase 2 - Unsupervised Discovery of Communities

In this phase, we will focus on the largest WCC to keep the analysis centered on the main body of the network. Tiny fragments can inflate the number of detected communities without contributing to answering the research question. For this analysis, detecting a single community result can be misleading if driven by the bias of a single algorithm. For this project, I compare several unsupervised methods:

- **Louvain**
- **Label Propagation**

Greedy Modularity was also considered as a potential unsupervised method, but it was computationally expensive and did not scale well with the size of the Twitter graph. Nonetheless, if multiple methods recover substantial community structure, then the conclusion that the graph contains tightly connected groups is strengthened.

```python
# ------------------------------------------------------
# Connectivity Structure
# ------------------------------------------------------

def communities_to_partition(communities):
  '''
  Convert a list of communities into a node for community ID mapping
  Useful for looking up which community a node belongs to
  '''
  part = {}
  # Assign a unique ID to each community
  for cid, community in enumerate(communities):
    for node in community:
      # Map node to its community
      part[node] = cid
  return part

# Run Louvain community detection using available implementation
def run_louvain(G_undirected):
  if HAVE_NX_LOUVAIN:
    # Use NetworkX built-in Lovauin if available
    communities = list(louvain_communities(G_undirected, seed=42))
    return [set(c) for c in communities]
  # Use python-louvain package (alternative implementation)
  elif HAVE_PYTHON_LOUVAIN:
    raw_partition = community_louvain.best_partition(G_undirected, random_state=42)
    # Convert node -> community mapping into grouped comunities
    grouped = defaultdict(set)
    for node, cid in raw_partition.items():
      grouped[cid].add(node)
    return list(grouped.values())
  else:
    return None

community_results = {}

# largest weakly connected component -> undirected graph
G_main_directed = graph.subgraph(largest_wcc).copy()
G_main = G_main_directed.to_undirected()

# Lovain on full graph
louvain_comms = run_louvain(G_main)
if louvain_comms is not None:
  community_results["Louvain"] = louvain_comms

# Label Propagation
community_results["Label Propagation"] = [set(c) for c in asyn_lpa_communities(G_main, seed=42)]

for method, communities in community_results.items():
    print(method, "->", len(communities), "communities")
```

# Phase 3 - Structural Isolation V.S Connectivity

# Community Quality Metrics

I have discovered a lot of communities, but that by itself is not enough. We need to test whether these communities are structurally coherent. To accomplish this, each method needs to be evaluated using several complementary metrics:

  - Modularity: measures how well the network is divided into groups (higher value = more separation)
  - Internal Edge Ratio: how many connections stay inside the same community
  - External Edge Ratio: how many connections go between different communities
  - Conductance: how isolated each community is from the rest of the network
  - Internal Density: how tightly connected the nodes are within each community

These metrics answer the following question: are the communities tightly conected internally while also being loosely connected to the rest of the network?

```python
def evaluate_partition(G_undefined, communities):
  '''
  Evaluate the quality of a community partition using several
  metrics

  '''
  partition = communities_to_partition(communities)
  m = G_undefined.number_of_edges()

  # Counters and lists for storing community-level metrics
  internal_edges = 0
  external_edges = 0
  conductances = []
  densities = []
  sizes = []

  # Count whether each edge stays inside a community or crosses communities
  for u, v in G_undefined.edges():
    if partition[u] == partition[v]:
      internal_edges += 1
    else:
      external_edges += 1

  # Compute conductance and density for each community
  for community in communities:
    community = set(community)
    sizes.append(len(community))
    # Skip singleton communities because their density/conductance is not meaningful
    if len(community) <= 1:
      continue

    # Internal density: how connected nodes are within the community
    sub = G_undefined.subgraph(community)
    internal_m = sub.number_of_edges()
    possible = len(community) * (len(community) - 1) / 2
    densities.append(internal_m / possible if possible > 0 else np.nan)

    # Conductance: how many edges leave the community relative to its volume
    volume_community = sum(dict(G_undefined.degree(community)).values())
    cut_edges = 0
    # Count edges from this community to outside nodes
    for node in community:
      for nbr in G_undefined.neighbors(node):
        if nbr not in community:
          cut_edges += 1
    volume_rest = 2 * G_undefined.number_of_edges() - volume_community
    denominator = min(volume_community, volume_rest)
    conductances.append(cut_edges / denominator if denominator > 0 else np.nan)

# Return summary for community detection method
  return {
    "Largest Community": max(sizes),
    "Median Community": float(np.median(sizes)),
    "Modularity": modularity(G_undefined, communities),
    "Internal Edge Ratio": internal_edges / m if m > 0 else np.nan,
    "External Edge Ratio": external_edges / m if m > 0 else np.nan,
    "Avg Conductance": float(np.nanmean(conductances)) if conductances else np.nan,
    "Median Conductance": float(np.nanmedian(conductances)) if conductances else np.nan,
    "Avg Internal Density": float(np.nanmean(densities)) if densities else np.nan,
    "Median Internal Density": float(np.nanmedian(densities)) if densities else np.nan
  }

# Evaluate each community detection method
eval_rows = []
for method, communities in community_results.items():
  row = evaluate_partition(G_main, communities)
  row["Method"] = method
  eval_rows.append(row)

eval_df = pd.DataFrame(eval_rows).sort_values("Modularity", ascending=False)
eval_df
```

The evaluation results show that both methods identify meaninful community structure, but they capture it in different ways.

Louvain achieves a higher modularity, indicating strong separation between communities and better overall clustering quality. It also produces fewer but larger communities. The internal edge ratio is high, meaning that most connections stay within communities and the external edge ratio is low, meaning that there is limited cross community interaction. A lower conductance means that most of the communities are isolated from the rest of the netwok.

On the other hand, Label Propagation produces a higher internal edge ratio and lower modularity, meaning that the partition is less structured globally. The higher conductance values indicate that the communities are more connected with the rest of the graph. Furthermore, Label Propagation has higher internal density, which suggests that the small communities it finds are tightly connected locally, even if they are not as well separated globally.

Overall, these results suggest that the network contains strong community structure at varying scales. Louvain captures larger, globally coherent communities, while Label Propagation reveals smaller, tightly connected local groups. This points to the conclusion that the network is organized into densely connected clusters with limited external connectivity, supporting the presence of echo-chamber-like structure.

# Internal V.S External Connectivity

If most edges remain within the detected communities rather than crossing between them, the network is organized into self-contained regions.

```python
edge_rows = []
for method, communities in community_results.items():
  partition = communities_to_partition(communities)
  internal = 0
  external = 0
  for u, v in G_main.edges():
    if partition[u] == partition[v]:
      internal += 1
    else:
      external += 1

  edge_rows.append({
      "Method": method,
      "Internal Edges": internal,
      "External Edges": external,
      "Internal Ratio": internal / (internal + external),
      "External Ratio": external / (internal + external)
  })

edge_df = pd.DataFrame(edge_rows).sort_values("Internal Ratio", ascending=False)
edge_df
```

```python
fig = plt.figure(figsize=(9, 5))
x = np.arange(len(edge_df))
width = 0.38
plt.bar(x - width/2, edge_df["Internal Ratio"], width=width, label="Internal Ratio")
plt.bar(x + width/2, edge_df["External Ratio"], width=width, label="External Ratio")
plt.xticks(x, edge_df["Method"], rotation=15)
plt.ylabel("Ratio")
plt.title("Internal vs External Edge Concentration")
plt.legend()
plt.show()
```

Both methods show high internal edge ratios and low external ratios, meaning that most connections stay within communities. This indicates strong, tightly clustered groups. Since both methods Produce similar patterns, it suggests the community structure is real and not dependent on a single algorithm.

# Conductance Distribution

Calculating the average conductance can be a useful metric and finding distributions only provides more detail. By plotting conductance values across communities, we can see whether insulation is widespread or concentrated in a specific subset.

```python
# Compute conductance values for each community in a partition
def conductance_values(G_undefined, communities):
  values = []
  # Loop through each detected community
  for community in communities:
    community = set(community)
    if len(community) <= 1:
      continue
    # Volume: total degree of all nodes inside community
    volume_community = sum(dict(G_undefined.degree(community)).values())
    # Count edges that leave the community
    cut_edges = 0
    for node in community:
      for nbr in G_undefined.neighbors(node):
        if nbr not in community:
          cut_edges += 1
    # Volume of rest of graph
    volume_rest = 2 * G_undefined.number_of_edges() - volume_community
    # Conductance uses smaller side of the cut as denominator
    denominator = min(volume_community, volume_rest)
    # Store conductance value for this community
    values.append(cut_edges / denominator if denominator > 0 else np.nan)
  return np.array(values, dtype=float)

# Plot conductance distributions for each community detection method
fig = plt.figure(figsize=(9,6))
for method, communities in community_results.items():
  values = conductance_values(G_main, communities)
  values = values[~np.isnan(values)]
  plt.hist(values, bins=30, alpha=0.5, label=method)

plt.xlabel("Conductance")
plt.ylabel("Frequency")
plt.title("Conductance Distributions")
plt.legend()
plt.show()
```

The Louvain communities tend to have low conductance, meaning they are more isolated and have fewer connections to other communities. On the other hand, Label Propagation shows higher and more spread out conductance values, indicating its communities are less isolated and more connected to the rest of the graph.

Overall, this suggests that Louvain produces more well-separated communities and Label Propagation captures smaller groups.

# Reduced Visualization of Community Structure

Here, I construct a reduced community-level network where each node represents an entire community and edges represent connections between communities. This allows for a clearer view of how communities interact and relate to one another at a higher level.

```python
# Louvain communities are main partition
viz_method = "Louvain"
viz_communities = community_results[viz_method]
viz_partition = communities_to_partition(viz_communities)

# Get community sizes
community_sizes = {
    cid: len(community)
    for cid, community in enumerate(viz_communities)
}

# Count edges between communiies
community_edges = defaultdict(int)
for u, v in G_main.edges():
  cu = viz_partition[u]
  cv = viz_partition[v]

  if cu != cv:
    pair = tuple(sorted((cu, cv)))
    community_edges[pair] += 1

# Build community-level graph
G_community = nx.Graph()

for cid, size in community_sizes.items():
  G_community.add_node(cid, size=size)

for (cu, cv), weight in community_edges.items():
  G_community.add_edge(cu, cv, weight=weight)

# Keep only largest communities for readability
top_k = 20
top_communities = sorted(
    community_sizes,
    key=community_sizes.get,
    reverse=True
)[:top_k]

G_community_viz = G_community.subgraph(top_communities).copy()

# Layout
pos = nx.spring_layout(G_community_viz, seed=42, weight='weight')

# Node sizes scaled by community size
node_sizes = [G_community_viz.nodes[n]['size'] / 10 for n in G_community_viz.nodes()]

# Edge widths scaled by cross-community connection strength
edge_widths = [G_community_viz[u][v]['weight'] / 1000 for u, v in G_community_viz.edges()]

plt.figure(figsize=(10,8))
nx.draw_networkx_nodes(G_community_viz, pos, node_size=node_sizes)
nx.draw_networkx_edges(G_community_viz, pos, alpha=0.15, width=edge_widths)
nx.draw_networkx_labels(G_community_viz, pos, font_size=10)

plt.title('Community-Level Network: Largest Louvain Communities')
plt.axis('off')
plt.show()
```

This visualization shows a community-level network of the largest Louvain communities. Each node represents an entire community, with larger nodes corresponding to larger communities. Edges between nodes represent connections between communities, with thicker edges indicating stronger cross-community connectivity.

It is evident that the network is not uniformly mixed. A small number of large communities dominate the structure and are more heavly connected to other communities. In contrast, smaller communities tend to be more peripheral and have fewer connections. This indicates that the connectivity is unevenly distributed across communities.

Furthermore, this visualization shows that the presence of edges between communities seems to suggest that there is still cross-community interaction. This aligns with the earlier findings showing high internal edge ratios nd relatively low condutance for Louvain communities.

# Phase 4 - The Role of Bridge Users

It is important to know whether some users connect to other separate communities. Strong communtieis do not imply aperfectly disconnected network. Some communities may remain dense while still relying on a set number of users to enable information flow across boundaries.

A bridge-user analysis helps refine this idea:

  - If there are no bridge users, the network is nearly disconnected
  - If there are many strong bridge users, then communities may be less insulated than they appear
  - If bridge capacity is concentrated in a small subset of users, then the network is clustered but selectively connected.

```python
full_partition = communities_to_partition(community_results[viz_method])

bridge_rows = []
for node in G_main.nodes():
  nbrs = list(G_main.neighbors(node))
  neighbor_communities = {full_partition[n] for n in nbrs if n in full_partition}
  bridge_rows.append({
      "Node": node,
      "Degree": G_main.degree(node),
      "Distinct Neighbor Communities": len(neighbor_communities)
  })

bridges_df = pd.DataFrame(bridge_rows).sort_values("Distinct Neighbor Communities", ascending=False)
bridges_df.head(20)
```

```python
fig = plt.figure()
plt.hist(bridges_df["Distinct Neighbor Communities"], bins=30)
plt.xlabel("Distinct Neighbor Communities")
plt.ylabel("Node Count")
plt.title("Distinct Neighbor Communities Distribution")
plt.show()

top_bridges = bridges_df.head(20)

```

The distribution of distinct neighbor communities is highly skewed as it shows that the majority of nodes connect to only a small amount of communities. This reveals that most users are primarily embedded within a single community or interact with only a few others.

At the same time, a small number of nodes connect to many different communities. These high-degree nodes act as bridge users, linking separate clusters. This shows that while communities are tightly knit, the network is not fully isolated and remains connected through a limited set of key nodes.



# Community-to-Community Edge Flow

A way to analyze bridge structure is to aggregate the graph into a community-level network. Each node in the reduced graph represents a community and weighted edges represent the number of cross-community links.

```python
community_flow = defaultdict(int)

# Iterate through all edges in graph
for u, v in G_main.edges():
  # Community IDs for each node
  cu = full_partition.get(u)
  cv = full_partition.get(v)
  # Only consider edges that go between different communities
  if cu is not None and cv is not None and cu != cv:
    pair = tuple(sorted((cu, cv)))
    community_flow[pair] += 1

# Convert community-to-community edge counts into a DataFrame
flow_df = pd.DataFrame([{"Community A": a, "Community B": b, "Cross Edges": w}
                        for (a,b), w in community_flow.items()
                        ]).sort_values("Cross Edges", ascending=False)

flow_df.head(20)
```

```python
if len(flow_df) > 0:
  fig = plt.figure()
  plt.hist(flow_df["Cross Edges"], bins=30)
  plt.xlabel("Cross-community Edge Count")
  plt.ylabel("community-pair Count")
  plt.title("Distribution of Cross-Community Connectivity")
  plt.show()
```

These results show that most community pairs have relatively few connections, which indicate that communities are well-separated. There is limited interaction between the edges, and this supports the idea of strong internal cohesion, where most edges remain within communities rather than spanning across them.

However, a small number of community pairs exhibit very high cross-edge counts. These pairs represent key pathways of interaction between major communities and suggest that connectivity is not evenly distributed across the network. These connections help keep the overall network connected, even though most users stay within their groups.

# Graph Embeddings as Community Structure

Community detection partitions the graph directly from connectivity patterns, but what if the same structure appears when nodes are embedded into a vector space.

Instead of working with only raw adjacency structure, we learn vector representations of nodes that preserve local and higher-order graph neighborhoods. If clustering in the embedding space produces coherent groups similar to graph-based communities, then the network contains meaningful latent structure rather thn only method-specifc partitions.

# Node2Vec Embeddings

Node2Vec learns node representations throguh random walks.

```python
!pip install node2vec
```

```python
import warnings
warnings.filterwarnings("ignore")

# Maximum number of nodes to use for embeddings
EMBED_SAMPLE_MAX_NODES = 5000

# build sample graph for embeddings if graph is too large
if G_main.number_of_nodes() > EMBED_SAMPLE_MAX_NODES:
  degree_sorted_nodes = [n for n, _ in sorted(G_main.degree(), key=lambda x: x[1], reverse=True)[:EMBED_SAMPLE_MAX_NODES]]
  # Create subgraph from top nodes
  G_embed = G_main.subgraph(degree_sorted_nodes).copy()
else:
  # If graph is small enough, use full graph
  G_embed = G_main.copy()

print("Embedding Graph Nodes:", G_embed.number_of_nodes())
print("Embedding Graph Edges:", G_embed.number_of_edges())

# Flag to check if Node2Vec ran successfully
node2vec_bool = False
try:
  from node2vec import Node2Vec
  node2vec = Node2Vec(
      G_embed,
      dimensions=64,
      walk_length=20,
      num_walks=50,
      workers=1,
      seed=42
  )
  node2vec_model = node2vec.fit(window=10, min_count=1, batch_words=128)

  # Get list of nodes used for embeddings
  embed_nodes = list(G_embed.nodes())
  # Build embeddings matrix
  X_embed = np.vstack([node2vec_model.wv[str(node)] for node in embed_nodes])
  # Mark Node2vec successful
  node2vec_bool = True
  print("Node2Vec Embeddings Learned Successfully")
except Exception as e:
  print("Node2Vec not available or failed to run:", e)
  print("skipping embedding section")
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score, normalized_mutual_info_score

# Run if Node2Vec embeddings were successfully createed
if node2vec_bool:
  # Get Louvain partition
  if "Louvain" in community_results:
    full_partition_louvain = communities_to_partition(community_results["Louvain"])

  # Create reference labels from Louvain communities
  embed_reference_labels = np.array([full_partition_louvain.get(n, -1) for n in embed_nodes])
  valid_mask = embed_reference_labels != -1

  # Number of clusters for KMeans
  k = 10

  # Run KMeans clustering on Node2Vec embeddings
  kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
  embed_cluster_labels = kmeans.fit_predict(X_embed)

  # Evaluation metrics
  # --------------------
  # Silhouette score: measures how well clusters are separated in embedding space
  sil = silhouette_score(X_embed, embed_cluster_labels)
  # ARI: compares clustering with Louvain communities
  ari = adjusted_rand_score(embed_reference_labels[valid_mask], embed_cluster_labels[valid_mask])
  # NMI: measures similarity between clustering
  nmi = normalized_mutual_info_score(embed_reference_labels[valid_mask], embed_cluster_labels[valid_mask])

  embed_eval_df = pd.DataFrame({
      "Metric": ["Embedding Silhouette", "ARI vs Graph Communities", "NMI vs Graph Communities", "Num Clusters"],
      "Value": [sil, ari, nmi, k]
  })

  display(embed_eval_df)
```

```python
if node2vec_bool:
  from sklearn.decomposition import PCA

  pca = PCA(n_components=2, random_state=42)
  X_2d = pca.fit_transform(X_embed)

  fig = plt.figure(figsize=(9,7))
  plt.scatter(X_2d[:, 0], X_2d[:, 1], s=10, c=embed_cluster_labels)
  plt.xlabel("PCA Component 1")
  plt.ylabel("PCA Component 2")
  plt.title("Node2Vec Embeddings Clustered with K-Means")
  plt.show()
```

The embedding-based clustering shows moderate alignment with the graph-based communities. The ARI and NMI indicate that the clusters found using Node2Vec and K-means are reasonably consistent with the Louvain communities but not exactly the same. This suggests that the structural patterns captured in the graph are also reflected in the learned embeddings.

It is worth meentioning that the silhouette score is low, which means that the clusters are not perfectly separated in the embedding space. Overall, this supports that the network contains real community structure even through different representations.

# Bridge Users as Structural Anomalies


```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Only create bridge_df if it doesn't already exist
if "bridge_df" not in globals():
  # Louvain community assignments
  partition_for_anomaly = communities_to_partition(community_results["Louvain"])
  bridge_rows = []
  for node in G_main.nodes():
    # Obtain neighbors of the node
    nbrs = list(G_main.neighbors(node))
    # Count different communities that the neighbors belong to
    neighbor_communities = {partition_for_anomaly[n] for n in nbrs if n in partition_for_anomaly}
    # Store structural features for each node
    bridge_rows.append({
        "Node": node,
        "Degree": G_main.degree(node),
        "Distinct Neighbor Communities": len(neighbor_communities)
    })
  bridge_df = pd.DataFrame(bridge_rows)

# Recompute partition
partition_for_anomaly = communities_to_partition(community_results["Louvain"])

bridge_df = bridge_df.copy()
# Approximate betweenness centrality
betweenness = nx.betweenness_centrality(
    G_main,
    k=min(200, G_main.number_of_nodes()),
    seed=42
)
# Map betweenness scores to nodes
bridge_df["Betweenness"] = bridge_df["Node"].map(betweenness)
# Compute custering coefficient
clustering_coefficient = nx.clustering(G_main)
# Map clustering coefficient to nodes
bridge_df["ClusteringCoefficient"] = bridge_df["Node"].map(clustering_coefficient)
# Store community ID for each node
bridge_df["CommunityID"] = bridge_df["Node"].map(partition_for_anomaly)

# Prepare features for anomaly detection
anomaly_features = bridge_df[["Degree", "Distinct Neighbor Communities", "Betweenness", "ClusteringCoefficient"]].fillna(0.0)

# Normalize features
scaler = StandardScaler()
X_anom = scaler.fit_transform(anomaly_features)

# Isolation Forest
iso = IsolationForest(contamination=0.02, random_state=42)
bridge_df["AnomalyLabel"] = iso.fit_predict(X_anom)
bridge_df["AnomalyScore"] = iso.decision_function(X_anom)

top_structural_anomalies = bridge_df.sort_values("AnomalyScore").head(20)
top_structural_anomalies[["Node", "Degree", "Distinct Neighbor Communities", "Betweenness", "ClusteringCoefficient", "AnomalyScore"]]
```

```python
fig = plt.figure(figsize=(8, 6))
mask_anom = bridge_df["AnomalyLabel"] == -1
plt.scatter(
    bridge_df.loc[~mask_anom, "Degree"],
    bridge_df.loc[~mask_anom, "Distinct Neighbor Communities"],
    s=12,
    label="Typical nodes"
)
plt.scatter(
    bridge_df.loc[mask_anom, "Degree"],
    bridge_df.loc[mask_anom, "Distinct Neighbor Communities"],
    s=20,
    label="Structural Anomalies"
)
plt.xlabel("Degree")
plt.ylabel("Distinct Neighbor Communities")
plt.title("Bridge Potential vs Degree with Anomaly Detection")
plt.legend()
plt.show()
```

The anomaly detection results suggest that bridge users are structurally unusual compared to most nodes in the network. Thje top anomalies tend to have very high degree, connect to many distinct communities, and have higher betweenness centrality. This means that they are well-connected users that play an important role in linking different parts of the graph.

Most typical nodes are concentrated at low degree and low numbers of distinct neighbor communities, meaning they stay within one local region of the network. In contrast, the anomaly nodes spread farther upward and to the right, showing that they connect to many communities and have broader reach. Overall, this indicates that most users are locally embedded, and a small set of unusual nodes act as important bridges between communities.

# Phase 5 - Robustness Through Randomized Baseline

One concern in network analysis is that strong communities might appear just because some nodes have many more connections than others. To check whether the communities we see are real, we compare the original graph to a randomized version that keeps the number of connections for each node but rewires how those connections are arranged.

If community structure is weaker after this randomization, then the original graph has meaningful structure and the communities are not just caused by how connected certain nodes are.

```python
G_random = G_main.copy()
# Determine numbre of edge swaps
# Rewire edges while preserving degree distribution
num_swaps = min(5 * G_random.number_of_edges(), G_random.number_of_edges())

try:
  # Double edge swaps to preserve degree of each node and randomize connections between nodes
  nx.double_edge_swap(G_random, nswap=num_swaps, max_tries=num_swaps * 20, seed=42)
  randomized_ok = True
except Exception as e:
  print("Randomization Warning:", e)
  randomized_ok = False

# Community detection on randomized graph
baseline_rows = []
if randomized_ok:
  baseline_results = {}

  # Louvain on randomized graph
  random_louvain = run_louvain(G_random)
  if random_louvain is not None:
    baseline_results["Louvain"] = random_louvain

  # Label Propagation on randomized graph
  baseline_results["Label Propagation"] = [set(c) for c in asyn_lpa_communities(G_random, seed=42)]

  for method, communities in baseline_results.items():
    row = evaluate_partition(G_random, communities)
    row["Method"] = method
    baseline_rows.append(row)

baseline_df = pd.DataFrame(baseline_rows)
baseline_df
```

```python
if randomized_ok and not baseline_df.empty:
  observed = eval_df.set_index("Method")
  baseline = baseline_df.set_index("Method")
  shared = sorted(set(observed.index) & set(baseline.index))

  compare_df = pd.DataFrame({
      "Observed Modularity": [observed.loc[m, "Modularity"] for m in shared],
      "Baseline Modularity": [baseline.loc[m, "Modularity"] for m in shared],
      "Observed Internal Ratio": [observed.loc[m, "Internal Edge Ratio"] for m in shared],
      "Baseline Internal Ratio": [baseline.loc[m, "Internal Edge Ratio"] for m in shared],
  }, index=shared)

  display(compare_df)

  fig = plt.figure(figsize=(9, 5))
  x = np.arange(len(shared))
  width = 0.38
  plt.bar(x - width/2, compare_df["Observed Modularity"], width=width, label="Observed")
  plt.bar(x + width/2, compare_df["Baseline Modularity"], width=width, label="Randomized Baseline")
  plt.xticks(x, shared, rotation=15)
  plt.ylabel("Modularity")
  plt.title("Observed vs Randomized Baseline")
  plt.legend()
  plt.show()
```

The results show that the original Twitter graph has much stronger community structure than the randomized graph.

For Louvain, the modularity drops from 0.803423 to about 0.130980. For Label Propagation, it drops even more. This means that once the edges are shuffled, the clear community structure mostly disappears and this reflects real organization in the network.

Furthermore, in the originl graph, both methods had high internal edge ratios and after randomization, Label Propagation's internal ratio becomes also 1.0 at the cost of its modularity. Overall, the original graph contains real, non-random clustering and this supports the existence of echo chambers.
