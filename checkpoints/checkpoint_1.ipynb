- Each dataset presents its own data quality issues that should be kept in mind when working with their data.

- Upon first glance, the first dataset is relatively clean structurally, but it is also fairly samll with only about 2,300 transactions/rows of data. This can limit statistical robustness and pattern generalizations. - Finally, the Yelp dataset includes substantial textual noise. Reviews may contain spelling errors, sarcasm, and contextual ambiguity. Additionally, rating imbalaqnce may affect model performances and interpretability.

# (D) Exploratory Data Analysis on Selected Dataset

```python
# ------------------------------------------------------
# 1. Download Directed Graph
# ------------------------------------------------------
# Source Page: https://snap.stanford.edu/data/ego-Twitter.html
# File name on SNAP: https://snap.stanford.edu/data/twitter_combined.txt.gz

import os, io, gzip, zipfile, tarfile, sys, math, random
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# download dataset
DATA_URL = "https://snap.stanford.edu/data/twitter_combined.txt.gz"
LOCAL_PATH = "data/twitter_combined.txt.gz"

os.makedirs("data", exist_ok=True)

def download_dataset(url: str, to_path: str):
    import urllib.request
    print(f"Downloading from {url} ...")
    urllib.request.urlretrieve(url, to_path)
    size = os.path.getsize(to_path) / (1024*1024)
    print(f"Saved to {to_path} ({size:.2f} MB)")

download_dataset(DATA_URL, LOCAL_PATH)
```

## Downloading Data

The ego-Twitter network dataset is obtained from the SNAP repository owned by Stanford. This directed grpah represents follower relationships between users.

```python
# ------------------------------------------------------
# 2. Load Directed Graph
# ------------------------------------------------------
import gzip

def load_directed_graph(path, comment='#', sep=None) -> nx.DiGraph:
  '''
  Load directed edge list into a DiGraph
  Assumes there are comment lines
  Assumes lines are formatted like: u<sep>v
  '''
  graph = nx.DiGraph()

  # line counters
  line_count = 0
  skipped_lines_count = 0

  open_fn = gzip.open if path.endswith('.gz') else open

  with open_fn(path, 'rt') as f:
    for line in f:
      line_count += 1

      # ignore lines that start with # (comments)
      if line.startswith(comment):
        skipped_lines_count += 1
        continue

      components = line.strip().split(sep)

      # ignore lines that are malformed
      if len(components) != 2:
        skipped_lines_count += 1
        continue

      try:
        src, dst = map(int, components)
        graph.add_edge(src, dst)
      except ValueError:
        skipped_lines_count += 1
        continue

    return graph

print('Loading graph...')
graph = load_directed_graph(LOCAL_PATH)
print('Graph loaded.')
```

## Loading Graph

The dataset is parsed into a NetworkX DiGraph which will be utilized for exploratory data analysis. As we parse through the data, we make sure to clean the data - remove comment lines or malformed entries. Edge endpoints ar econverted to integer form for efficiency and consistency purposes.

```python
# ------------------------------------------------------
# 3. Basic Graph Statistics
# ------------------------------------------------------

# compute statistics
num_nodes = graph.number_of_nodes()
num_edges = graph.number_of_edges()
density = nx.density(graph)
reciprocity = nx.reciprocity(graph)
transitivity = nx.transitivity(graph)
num_self_loops = nx.number_of_selfloops(graph)

# ---- Print Statements ----
print('Nodes:', num_nodes)
print('Edges:', num_edges)
print('Density:', density)
print('Reciprocity:', reciprocity)
print('transitivity:', transitivity)
print('Number of Self-loops:', num_self_loops)
print('Directed:', nx.is_directed(graph))
```

## Basic Graph Statistics

Here, we compute foundational structural metrics: node count, edge count, density, reciprocity, transitivity, and self loops. These computational statistics provide insight into the graph's scale and sparsity.

```python
# ------------------------------------------------------
# 4. Degree Analysis (In/Out)
# ------------------------------------------------------

in_degrees = pd.Series(dict(graph.in_degree()), name='in_degree')
out_degrees = pd.Series(dict(graph.out_degree()), name='out_degree')

degree_df = pd.concat([in_degrees, out_degrees], axis=1)
print('Degree Statistics')
degree_df = degree_df.describe(percentiles=[0.5, 0.9, 0.95, 0.99])
degree_df = degree_df.rename(index={'50%': 'median'})
display(degree_df)

# plot histogram: in-degree + out-degree
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(in_degrees, bins=100)
plt.yscale('log') # log scale frequency
plt.title('In-Degree Distribution')
plt.xlabel('In-Degree')
plt.ylabel('Frequency (log scale)')

plt.subplot(1,2,2)
plt.hist(out_degrees, bins=100)
plt.yscale('log') # log scale frequency
plt.title('Out-Degree Distribution')
plt.xlabel('Out-Degree')
plt.ylabel('Frequency (log scale)')
plt.tight_layout()
plt.show()
```

## Degree Analysis

We analyze in-degree and out-degree distributions of the graph to gain insight into the nature of the network, specifically the roles within the graph. The in-degrees represent the popularity while the out-degree represents the activity of a user. We utilize plots to visualize how these distributions vary and to confirm the typical behavior of this social network.

```python
# ------------------------------------------------------
# 5. Component Structure Analysis
# ------------------------------------------------------

# compute components
sccs = list(nx.strongly_connected_components(graph))
wccs = list(nx.weakly_connected_components(graph))

# sort by size
scc_sizes = sorted([len(c) for c in sccs], reverse=True)
wcc_sizes = sorted([len(c) for c in wccs], reverse=True)

# largest WCC/SCC size
largest_scc_size = scc_sizes[0]
largest_wcc_size = wcc_sizes[0]

# ---- Print Statements ----
print('Number of SCCs:', len(sccs))
print('Number of WCCs:', len(wccs))
print('\nTop 5 SCC sizes:', scc_sizes[:5])
print('Top WCC size:', wcc_sizes)
print('\nFraction of nodes in largest SCC:', largest_scc_size / graph.number_of_nodes())
print('Fraction of nodes in largest WCC:', largest_wcc_size / graph.number_of_nodes())
```

## Connectivity Structure

SCCs and WCCs are analyzed to understand mutual connectivity and reachability. With SCCs, we can reveal directional cycles and interconnected cores. On the other hand, with WCCs, can see a more general global structure.

```python
# ------------------------------------------------------
# 6. Approximate Shortest Path Sampling
# ------------------------------------------------------
import random

# extract largest WCC
largest_wcc_node = max(nx.weakly_connected_components(graph), key=len)
g_subgraph = graph.subgraph(largest_wcc_node).copy()

# convert to undirected for path length analysis
# directed paths restrict movement
g_subgraph_undirected = g_subgraph.to_undirected()

# confirm largest wcc node - should be the entire graph
print('Largest WCC Nodes:', g_subgraph.number_of_nodes())
print('Largest WCC Edges:', g_subgraph.number_of_edges())

# approximate shortest path sampling with 100 nodes
sample_nodes = random.sample(list(g_subgraph_undirected.nodes()), 100)

path_lengths = []
for node in sample_nodes:
  lengths = nx.single_source_shortest_path_length(g_subgraph_undirected, node)
  path_lengths.extend(lengths.values())

average_path_length = np.mean(path_lengths)
max_math_length = np.max(path_lengths)

print('\nEstimated Average Shortest Path Length:', average_path_length)
print('Maximum Average Shortest Path Length:', max_math_length)

# Path Length Distribution Plot
plt.figure()
plt.hist(path_lengths, bins=20)
plt.title('Estimated Shortest Path Length Distribution')
plt.xlabel('Path Length')
plt.ylabel('Count')
plt.show()
```

## Approximate Shortest Path Sampling

To evaluate global navigation, we utilize shortest path lengths by sampling on the largest weakly connected component. The graph is converted to undirected form to avoid path length inflation and provide more accurate structural cohesion analysis. The graph visualizes which path lengths are most common.

```python
# ------------------------------------------------------
# 7. Community Structure (Modularity)
# ------------------------------------------------------
import community as community_louvain

louvain_graph = g_subgraph.to_undirected()
partition = community_louvain.best_partition(louvain_graph)
num_communities = len(set(partition.values()))

print('Number of Communities:', num_communities)
```

## Community Structure

Utilizing community_louvain, we are able to reduce the amount of computational constraints involved when detecting communities in larger graphs. This provides insight into clustering patterns while maintaining feasibility on a larger network.
